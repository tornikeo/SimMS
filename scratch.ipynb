```python
import matchms as mms


gnps = tuple(mms.importing.load_from_mgf('data/downloads/GNPS-random-10k.mgf'))
```

```python
from matchms.filtering import reduce_to_number_of_peaks

gnps = map(lambda sp: reduce_to_number_of_peaks(sp, n_max=1024), gnps)
gnps = [sp for sp in gnps if sp is not None]
```

```python
len(gnps)
```

```python
import numpy as np

spectra = gnps
dtype = np.float32
n_max_peaks = 1024

mz = np.zeros((len(spectra), n_max_peaks), dtype=dtype)
int = np.zeros((len(spectra), n_max_peaks), dtype=dtype)
spectra_lens = np.zeros(len(spectra), dtype=np.int32)
```

```python
for i, s in enumerate(spectra):
  if s is not None:
    spec_len = min(len(s.peaks), n_max_peaks)
    mz[i, :spec_len] = s._peaks.mz[:spec_len]
    int[i, :spec_len] = s._peaks.intensities[:spec_len]
    spectra_lens[i] = spec_len
```

```python
mz
```

```python
np.save('data/gnps-mz.npy', mz)
np.save('data/gnps-int.npy', int)
np.save('data/gnps-spectra_lens.npy', spectra_lens)
```
